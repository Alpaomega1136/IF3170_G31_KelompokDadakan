# SVM Scratch dengan Preprocessing Mandiri

Satu SVM RBF dari scratch dengan preprocessing yang dihitung sendiri menggunakan NumPy, pandas, dan fungsi matematika SciPy. Estimasi mutual information mengikuti Ross (2014); pemilihan working set SVM mengacu pada Fan, Chen, dan Lin (2005). SVC hanya digunakan sebagai pembanding validasi.

## Persiapan

Pilih kernel `.venv`, lalu jalankan semua sel secara berurutan dari kernel baru. Pelatihan dapat memakan beberapa menit; versi paket ada di `requirements-svm.txt`.

```python
from pathlib import Path
import sys
candidates = [Path.cwd(), *Path.cwd().parents]
root = next((p for p in candidates if (p / 'data/train.csv').is_file() and (p / 'src/svm_safe_submission.py').is_file()), None)
if root is None:
    root = next((p / 'IF3170_G31_KelompokDadakan' for p in candidates if (p / 'IF3170_G31_KelompokDadakan/data/train.csv').is_file()), None)
if root is None:
    raise FileNotFoundError('Buka notebook dari folder proyek atau folder induknya')
sys.path.insert(0, str(root / 'src'))
import svm_safe_submission as svm
X, y, X_test, test, sample = svm.load_data()
print('Train:', X.shape, 'Test:', X_test.shape)
print('Konfigurasi:', svm.CONFIG)
```

## Validasi

Pisahkan 20% data berlabel untuk validasi; semua statistik dan bobot fitur hanya di-fit pada 80% sisanya. Angka validasi ini ikut dilihat saat memilih konfigurasi sehingga bukan estimasi independen sepenuhnya. Scikit-learn digunakan pada sel ini untuk split, metrik, dan model pembanding.

```python
from IPython.display import display
hasil = svm.validate(X, y)
display(hasil.round(5))
```

## Buat submission

Latih ulang preprocessing dan SVM scratch pada seluruh data berlabel. Data test hanya ditransformasi dan diprediksi; tidak ada label test yang digunakan. Model dan preprocessing disimpan, dimuat ulang, dan hasil prediksi diperiksa.

```python
submission, metadata = svm.make_submission(X, y, X_test, test, sample)
display(submission.head())
print(metadata)
```

Kandidat baru untuk kompetisi: `output/SVM/safe/submission_svm_safe.csv`, dengan threshold -0,2. File `submission_svm_safe_baseline.csv` memakai threshold -0,3 sebagai pembanding. Keduanya memakai model yang sama. Skor Kaggle kandidat baru baru diketahui setelah diunggah.
