# SVM: Eksperimen Bobot Fitur dan Submission

Notebook ini melatih satu SVM RBF dari scratch menggunakan NumPy. Imputasi, standardisasi, one-hot encoding, dan bobot mutual information dilatih hanya pada data latih. SVC scikit-learn digunakan sebagai pembanding validasi.

## Persiapan

Pilih kernel Python dari `.venv`, lalu jalankan semua sel berurutan dari kernel baru. Pelatihan dapat memakan waktu beberapa menit. Versi paket tercatat dalam `requirements-svm.txt`.

```python
from pathlib import Path
import sys

candidates = [Path.cwd(), *Path.cwd().parents]
root = next((p for p in candidates if (p / 'data/train.csv').is_file() and (p / 'src/svm_pro.py').is_file()), None)
if root is None:
    root = next((p / 'IF3170_G31_KelompokDadakan' for p in candidates if (p / 'IF3170_G31_KelompokDadakan/data/train.csv').is_file()), None)
if root is None:
    raise FileNotFoundError('Buka notebook dari folder proyek atau folder induknya')
sys.path.insert(0, str(root / 'src'))
import svm_pro as svm
X, y, X_test, test, sample = svm.common.load_data()
print('Train:', X.shape, 'Test:', X_test.shape)
print('Konfigurasi:', svm.CONFIG)
```

## Validasi

Data berlabel dibagi secara stratified: 80% untuk melatih preprocessor dan model, 20% untuk validasi. Threshold −0,3 mengikuti konfigurasi submission sebelumnya. Split ini ikut dilihat saat memilih kandidat; angkanya dapat optimistis dan bukan skor Kaggle.

```python
from IPython.display import display
hasil = svm.validate(X, y)
display(hasil.round(5))
```

## Latih ulang dan buat submission

Preprocessor dan SVM scratch dilatih ulang pada seluruh `train.csv`. `test.csv` hanya ditransformasi dan diprediksi. Model dan preprocessor disimpan dan dimuat ulang untuk memeriksa prediksi.

```python
submission, metadata = svm.make_submission(X, y, X_test, test, sample)
display(submission.head())
print(metadata)
```

CSV untuk kompetisi: `output/SVM/pro/submission_svm_pro.csv`. Folder yang sama berisi hasil validasi, konfigurasi, model scratch, dan preprocessor. Skor Kaggle baru diketahui setelah CSV diunggah.
